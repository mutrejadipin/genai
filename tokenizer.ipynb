```python
text="hello World!"
```

```python
import tiktoken
tokenizer=tiktoken.encoding_for_model(model_name="gpt-4")
token_ids=tokenizer.encode(text)
print(token_ids)
```

```python
result_ids=[4435,0,15339]
tokenizer.decode(result_ids)
```
